# ABC Company Employee Data Analysis

This notebook contains preprocessing, analysis, and visualization of ABC Company’s employee dataset.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files
uploaded = files.upload()

df = pd.read_excel("ABC Company.xlsx")
df.head()
```

##  Preprocessing
Replace the "height" column with random values between 150–180.


```python
df['height'] = np.random.randint(150, 181, size=len(df))
df['height'].head()
```

## Distribution of Employees Across Teams


```python
team_counts = df['Team'].value_counts()
team_percent = (team_counts / len(df)) * 100

plt.figure(figsize=(12,10))
team_counts.plot(kind='pie',autopct='%1.0f%%')
plt.title("Employee Distribution by Team")
plt.show()
```

##  Segregate Employees by Position


```python
position_counts = df['Position'].value_counts()

plt.figure(figsize=(8,6))
position_counts.plot(kind='bar', color='skyblue')
plt.title("Employees by Position")
plt.show()
```

##  Predominant Age Group


```python
age_bins = [20, 30, 40, 50, 60]
df['age_group'] = pd.cut(df['Age'], bins=age_bins)
age_group_counts = df['age_group'].value_counts()

df['Age'].plot(kind='hist', bins=age_bins, color='orange', edgecolor='black')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.show()
```

## Salary Expenditure by Team & Position


```python
salary_by_team = df.groupby('Team')['Salary'].sum()
salary_by_position = df.groupby('Position')['Salary'].sum()

plt.figure(figsize=(8,6))
salary_by_team.plot(kind='bar', color='green')
plt.title("Salary Expenditure by Team")
plt.show()

plt.figure(figsize=(8,6))
salary_by_position.plot(kind='bar', color='purple')
plt.title("Salary Expenditure by Position")
plt.show()
```

## Correlation Between Age and Salary


```python
sns.regplot(x='Age', y='Salary', data=df, scatter=True)
plt.title("Age vs Salary Trend")
plt.show()
```

## Insights & Data Story
- Team distribution shows which teams dominate in size.
- Position segregation highlights the most common roles.
- Age distribution reveals the predominant age group.
- Salary expenditure analysis shows which team and position drive costs.
- Age vs Salary correlation provides insight into compensation trends.
